# Part A: Python Foundations


```python
# ============================================================
# Task 1-5: Foundations
# ============================================================

# Store and display total number of developers
Total_developers = len(developers)
print(f"Total number of developers in the system: {Total_developers}")

# Ask user to input a team name and display its members
Team_input = input("Enter a team name to view its members (e.g. Backend, Frontend, DevOps): ").strip()

team_members = [d for d in developers if d["team"].lower() == Team_input.lower()]
if team_members:
    print(f"\nDevelopers in '{Team_input}' team:")
    for dev in team_members:
        print(f"  - {dev['name']} (ID: {dev['id']}, Tasks: {dev['tasks']}, Score: {dev['score']})")
else:
    print(f"No team named '{Team_input}' found in the system.")

# Display average code quality score
Team_score = sum(dev["score"] for dev in developers)
avg_score = Team_score / Total_developers
print(f"\nAverage Code Quality Score across all developers: {avg_score:.2f}")

# Check if active developers > inactive developers
active_developers = 0
inactive_developers = 0

for dev in developers:
    if dev["tasks"] > 5:
        active_developers += 1
    else:
        inactive_developers += 1

print(f"\nActive developers  (tasks > 5): {active_developers}")
print(f"Inactive developers(tasks ≤ 5): {inactive_developers}")

if active_developers > inactive_developers:
    print("Active developers outnumber inactive developers.")
elif active_developers < inactive_developers:
    print("Inactive developers outnumber active developers.")
else:
    print("Active and inactive developers are equal.")
    
```

# Part B: Data Structures & Control Flow

```python
# ============================================================
# Task 6: Store unique teams using a set
# ============================================================

teams = set(dev["team"] for dev in developers)
print("Unique Teams in the system:", teams)

# ============================================================
# Task 7: Display first 5 developer names
# ============================================================

print("\nFirst 5 Developers:")
for d in developers[:5]:
    print(f"  - {d['name']} (ID: {d['id']}, Team: {d['team']}, Tasks: {d['tasks']}, Score: {d['score']}, Year: {d['year']})")


```

# Part C: Functions & Logic

```python
# ============================================================
# Task 11: Function — average score per team
# ============================================================

def average_score_per_team(developers):
    team_scores = {}
    for dev in developers:
        team = dev["team"]
        score = dev["score"]
        if team not in team_scores:
            team_scores[team] = []
        team_scores[team].append(score)
    
    avg_scores = {team: sum(scores)/len(scores) for team, scores in team_scores.items()}
    return avg_scores

print("Average Score Per Team:")
for team, avg in average_score_per_team(developers).items():
    print(f"  {team:<12}: {avg:.2f}")
```

```python
# ============================================================
# Task 12:  Function — top-performing developer per team
# ============================================================

def get_top_performers(developers):
    top_performers = {}
    for dev in developers:
        team = dev["team"]
        if team not in top_performers or dev["score"] > top_performers[team]["score"]:
            top_performers[team] = {"name": dev["name"], "score": dev["score"]}
    return top_performers

print("Top Performer Per Team:")
for team, info in get_top_performers(developers).items():
    print(f"  {team:<12}: {info['name']} (Score: {info['score']})")
```

```python
# ============================================================
# Task 13: Function — developers by joining year
# ============================================================

def developers_by_year(developers):
    year_dict = {}
    for dev in developers:
        year = dev["year"]
        if year not in year_dict:
            year_dict[year] = []
        year_dict[year].append(dev["name"])
    return year_dict

print("Developers by Joining Year:")
for year, names in sorted(developers_by_year(developers).items()):
    print(f"  {year}: {', '.join(names)}")
```

```python
# ============================================================
# Task 14: Function — performance category based on score
#   score < 50     → Poor
#   50 ≤ score ≤ 75→ Average
#   score > 75     → Excellent
# ============================================================

def performance_categories(developers):
    categories = {"Poor": [], "Average": [], "Excellent": []}
    for dev in developers:
        if dev["score"] < 50:
            categories["Poor"].append(dev["name"])
        elif dev["score"] <= 50 or dev["score"] <= 75:
            categories["Average"].append(dev["name"])
        else:
            categories["Excellent"].append(dev["name"])
    return categories

print(f"{'Name':<12} {'Score':>6}  {'Category'}")
print("-" * 32)
best = performance_categories(developers)
for best, names in best.items():
    for name in names:
        dev = next(d for d in developers if d['name'] == name)
        print(f"  {dev['name']:<10} {dev['score']:>6}  {best}")
```

```python
# ============================================================
# Task 15: Nested loop report — developers per team per category
# ============================================================

def nested_dict(developers):
    nested = {}
    for dev in developers:
        team = dev["team"]
        if team not in nested:
            nested[team] = []
        nested[team].append({"name": dev["name"], "score": dev["score"]})
    return nested

def performance_category(score):
    if score < 50:
        return "Poor"
    elif score <= 75:
        return "Average"
    else:
        return "Excellent"

print("=" * 50)
print("  Developers Per Team Per Performance Category")
print("=" * 50)

for team, members in sorted(nested_dict(developers).items()):
    print(f"\n  Team: {team}")
    for dev in ["Poor", "Average", "Excellent"]:
        in_dev = [m["name"] for m in members if performance_category(m["score"]) == dev]
        if in_dev:
            print(f"    [{dev}]: {', '.join(in_dev)}")
        else:
            print(f"    [{dev}]: None")

print("=" * 50)
```

# Part D: Error Handling 

```python
# ============================================================
# Task 16: Safely accept developer ID input
# ============================================================

def find_developer(dev_id):
    for d in developers:
        if d["id"] == dev_id:
            return d
    raise ValueError(f"Developer with ID {dev_id} does not exist.")

try:
    user_id = input("Enter Developer ID to look up: ").strip()

    if not user_id.isdigit():
        raise ValueError("Invalid ID")

    user_id = int(user_id)
    dev = find_developer(user_id)
    print(f" Developer found: {dev['name']} | Team: {dev['team']} | "
          f"Tasks: {dev['tasks']} | Score: {dev['score']}")

except Exception as e:
    print(f" Error: {e}")

finally:
    print("Developer lookup process completed")
```

# Part E: Developer Ranking Game 

```python
# ============================================================
# Task 17: Calculate final score for each developer
# ============================================================

def final_score_developer(developers):
    for dev in developers:
        dev["final_score"] = dev["tasks"] * 5 + dev["score"]
    return developers

final_score_developer(developers)

print(f"{'Name':<12} {'Tasks':>6}  {'Quality':>8}  {'Final Score':>12}")
print("-" * 45)
for dev in developers:
    print(f"  {dev['name']:<10} {dev['tasks']:>6}  {dev['score']:>8}  {dev['final_score']:>12}")

```

```python
# ============================================================
# Task 18: Display leaderboard (highest to lowest final score)
# ============================================================

def get_score(dev):
    return dev["score"]

def leaderboard(developers):
    sorted_devs = sorted(developers, key=get_score, reverse=True)
    return sorted_devs[:3]

print("Top 3 by Code Quality Score:")
for i, dev in enumerate(leaderboard(developers), start=1):
    print(f"  {i}. {dev['name']} (Score: {dev['score']})")
```

```python
# ============================================================
# Task 19: Declare best-performing developer
# ============================================================

def best_developer(developers):
    best = developers[0]  # assume first is best initially

    for d in developers:
        if d["final_score"] > best["final_score"]:
            best = d

    print("Best Developer:", best["name"])
    return best

best = best_developer(developers)
print("Best Performing Developer of the Season:")
print(f"   Name      : {best['name']}")
print(f"   Team      : {best['team']}")
print(f"   Tasks     : {best['tasks']}")
print(f"   Quality   : {best['score']}")
print(f"   Final Score: {best['final_score']}")
```

```python
# ============================================================
# Task 20: Head-to-head battle — compare two developers
# ============================================================

def battle_mode(dev_list):
    """Prompt for two developer IDs, compare, and declare a winner."""
    print("\n Head-to-Head Battle Mode")
    print("Available IDs:", [d["id"] for d in dev_list])

    try:
        id1 = int(input("  Enter first Developer ID : ").strip())
        id2 = int(input("  Enter second Developer ID: ").strip())

        d1 = next((d for d in dev_list if d["id"] == id1), None)
        d2 = next((d for d in dev_list if d["id"] == id2), None)

        if not d1 or not d2:
            print("  Developer not found")
            return

        print(f"\n  {d1['name']} (Score: {d1['final_score']})  vs  "
              f"{d2['name']} (Score: {d2['final_score']})")

        if d1["final_score"] > d2["final_score"]:
            print(f"   Winner: {d1['name']}")
        elif d2["final_score"] > d1["final_score"]:
            print(f"   Winner: {d2['name']}")
        else:
            print("   It's a TIE!")

    except ValueError:
        print("  Invalid input. Please enter numeric IDs.")


# call
battle_mode(developers)
```

```python
# ============================================================
# Task 21: Assign levels based on final score
#   final_score < 80   → Bronze
#   80 ≤ score ≤ 120   → Silver
#   score > 120        → Gold
# ============================================================

def assign_level(final_score):
    """Returns the badge level for a given final score."""
    if final_score < 80:
        return "Bronze"
    elif final_score <= 120:
        return "Silver"
    else:
        return "Gold"


print("=" * 50)
print("          Developer Badge Assignments")
print("=" * 50)
print(f"  {'Name':<12} {'Final Score':>12}  {'Level'}")
print("-" * 40)
for dev in developers:
    level = assign_level(dev["final_score"])
    print(f"  {dev['name']:<12} {dev['final_score']:>12}  {level}")
print("=" * 50)
print("\n Developer Performance Analytics")
```
